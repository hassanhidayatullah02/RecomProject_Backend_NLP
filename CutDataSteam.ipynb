```python
import pandas as pd

print("Membaca dataset original...")
steam = pd.read_csv('Data/steam.csv')
desc = pd.read_csv('Data/steam_description_data.csv')
media = pd.read_csv('Data/steam_media_data.csv')

# 1. Isi nilai kosong dengan string kosong di kedua kolom
steam['genres'] = steam['genres'].fillna('')
# steam['steamspy_tags'] = steam['steamspy_tags'].fillna('') # Kolom rahasianya ada di sini

# 2. Cari kata 'Horror' di kolom steamspy_tags ATAU genres
# kondisi_genre = steam['genres'].str.contains('Horror', case=False)
kondisi_tag = steam['steamspy_tags'].str.contains('Horror', case=False)

# Gunakan simbol | (OR) untuk menggabungkan dua kondisi
steam_horror = steam[kondisi_tag]
# steam_horror = steam[kondisi_genre]

# 3. Ambil daftar 'appid'
kumpulan_id_horror = steam_horror['appid'].tolist()

print(f"Ditemukan {len(kumpulan_id_horror)} game horror.")
print("Memangkas dataset deskripsi dan media...")

# 4. Potong data lainnya
desc_horror = desc[desc['steam_appid'].isin(kumpulan_id_horror)]
media_horror = media[media['steam_appid'].isin(kumpulan_id_horror)]

# 5. Simpan ke CSV baru
print("Menyimpan ke CSV baru...")
steam_horror.to_csv('Data/steam_horror.csv', index=False)
desc_horror.to_csv('Data/steam_desc_horror.csv', index=False)
media_horror.to_csv('Data/steam_media_horror.csv', index=False)

print("Selesai! File CSV khusus horror berhasil dibuat.")
```
